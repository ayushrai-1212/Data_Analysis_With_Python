# 1. Series and DataFrame

* Series : one column of values with labels (1D)
* DataFrame : a full table of rows and columns (2D). Every column of a DataFrame is a Series.

```python
import pandas as pd
import numpy as np     # numpy is used for missing values (np.nan), same library from Unit 2
```

### Series

```python
ls=[72,85,90,66]
sd=pd.Series(ls,index=["Amit","Riya","Sam","Neha"])
print(sd)
```

```python
print(sd["Riya"])    # access using label
print(sd.mean())     # statistics work directly on a series
sd+5                 # operation on all values at once, no loop needed
```

### DataFrame

```python
stu=pd.DataFrame({
    "Name":["Amit","Riya","Sam","Neha"],
    "Branch":["CSE","ECE","CSE","ME"],
    "Marks":[72,85,90,66]
})
stu       # keys of the dictionary become column names
```

```python
print(type(stu))
print(type(stu["Marks"]))   # one column of a dataframe is a series
```

```python
nd=stu["Marks"].values      # .values gives the numpy array inside a series
print(nd)
type(nd)
```

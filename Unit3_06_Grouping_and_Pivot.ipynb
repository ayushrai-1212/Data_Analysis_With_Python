# 3. Filtering Rows (Boolean Masking)

Filtering means keeping only the rows where a condition is True.

1. A condition like `df["Profit"]<0` gives True/False for every row. This is called a mask.
2. `df[mask]` keeps only the True rows.

| Operator | Meaning |
|---|---|
| `==`  `!=`  `>`  `>=`  `<`  `<=` | compare |
| `&` | AND (both conditions true) |
| `\|` | OR (any one true) |
| `~` | NOT |

Remember: `==` is for comparing, `=` is for assigning. Use `&` `|` and not `and` `or`. Each condition must be inside brackets.

```python
df["Profit"]<0      # this is the mask, True where the order made loss
```

```python
big_orders=df[df["Sales"]>1000]     # only True rows are kept
print(big_orders.shape)
big_orders.head()
```

```python
df[df["Region"]=="Central"].head()
```

### More than one condition

```python
df[(df["Profit"]<0) & (df["Category"]=="Technology")].head()
```

```python
df[(df["Region"]=="West") & (df["Quantity"]>=10)].head()
```

```python
# OR : same day delivery or very high discount
df[(df["Ship Mode"]=="Same Day") | (df["Discount"]>=0.5)].shape
```

# 4. Indexing: .loc and .iloc

Every dataframe has an index (row labels). By default it is 0,1,2,...

| | .loc | .iloc |
|---|---|---|
| uses | labels (index name, column name) | position (0,1,2...) |
| end of slice | included | not included (like python list) |
| boolean condition | can be used | not used |

A boolean condition can be used inside `.loc` to filter rows and choose columns in one step: `df.loc[condition,[columns]]`

```python
df.iloc[0:5]       # rows at position 0 to 4, 5 is not included
```

```python
df.iloc[0:3,0:4]   # first 3 rows and first 4 columns
```

```python
df.loc[0:4,["Order ID","Sales"]]    # rows with label 0 to 4, here 4 is included
```

```python
df.loc[(df['Sales']>=7000)&(df['Category']=='Technology'),['Row ID','Customer ID','Customer Name','Category','Sales']]
```

```python
df.loc[(df['Profit']<0)& (df['Category']=='Furniture'),['Order ID','Category','Sales','Discount','Profit']].head()
```

### set_index() and reset_index()

`set_index` makes a column the row labels. We keep the result in a new variable `df1` so `df` stays same.

```python
df1=df.set_index('Row ID')    # used for setting a particular column as index column
df1.head(3)
```

```python
print(df1.loc[11].name)      # label 11 means row with Row ID 11
print(df1.iloc[11].name)     # position 11 is the 12th row, so Row ID is 12
```

```python
df1.reset_index().head(3)    # index comes back as a normal column
```

# 6. Grouping and Pivot Tables

`groupby` works in 3 steps: split the rows into groups, apply a function (sum, mean...) on each group, combine the result.

`df.groupby("group column")["value column"].function()`

```python
df.groupby('Region')['Profit'].sum()
```

```python
df.groupby('Category')[['Profit','Sales']].mean()     # list for more than one column
```

```python
cat=df.groupby("Category")["Sales"].sum()
cat
```

```python
cat.sort_values(ascending=False)       # sort_values to rank the groups
```

```python
df.groupby("Region")["Sales"].sum().idxmax()     # name of the region with highest sales
```

```python
# two group columns, one result for every Region and Category pair
df.groupby(["Region","Category"])["Profit"].sum().reset_index()    # reset_index brings group names back as columns
```

```python
# count is number of rows, nunique is number of different orders. They are not same
df.groupby("Category")["Order ID"].agg(["count","nunique"])
```

### agg(): many calculations at once

* list : `.agg(["mean","count"])` same functions for all columns
* dictionary : `.agg({"Sales":"sum","Profit":"mean"})` different function for each column
* named : `.agg(Total_Sales=("Sales","sum"))` we choose the name of output column

```python
df.groupby("Category")[["Profit","Sales"]].agg(['mean','count','max'])   ## this function is used to fetch multiple statistical values
```

```python
df.groupby("Region").agg({"Sales":"sum","Profit":"mean"})
```

```python
df.groupby("Region").agg(
    Total_Sales=("Sales","sum"),
    Avg_Profit=("Profit","mean")
)
```

Top N rows inside each group: sort first, then `groupby().head(n)` keeps first n rows of every group.

```python
# top 2 profit transactions of every Region
df.sort_values("Profit",ascending=False).groupby("Region").head(2)[["Region","Order ID","Profit"]]
```

### pivot_table()

Same as groupby but shown as a grid. `index` is rows, `columns` is columns, `values` is what to calculate, `aggfunc` is how. `margins=True` adds a row and column named All with the totals.

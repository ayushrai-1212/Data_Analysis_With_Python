#Electricity bill calculator

100 units 5 rupees
200 units 7 rupees
350 10 rupees
450 15 rupees

````python
units = int(input("Enter the number of units consumed: "))
bill = 0

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
elif units <= 350:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)
elif units <= 450:
    bill = (100 * 5) + (100 * 7) + (150 * 10) + ((units - 350) * 15)
else:
    bill = (100 * 5) + (100 * 7) + (150 * 10) + (100 * 15) + ((units - 450) * 20) # Assuming 20 for units above 450
print(f"Your electricity bill is: ${bill}")
````

### Q1. Right-Angled Triangle (Asterisks)

Print a right-angled triangle pattern of asterisks:

```
*
**
***
****
```

````python
rows = int(input("Enter the no of rows you want:- "))
for i in range(1, rows + 1):
    for j in range(i):
        print('*', end='')
    print()
````

### Q2. Pyramid (Asterisks)

Print a pyramid pattern of asterisks:

```
   *
  ***
 *****
*******
```

````python
rows = int(input("Enter the no of rows you want:- "))
for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
````

### Q3. Diamond (Asterisks)

Print a diamond pattern of asterisks:

```
   *
  ***
 *****
*******
 *****
  ***
   *
```

````python
rows = int(input("Enter the no of rows you want:- "))
# Upper part of the diamond
for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
# Lower part of the diamond
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
````

### Q4. Hollow Square (Asterisks)

Print a hollow square pattern of asterisks:

```
*****
*   *
*   *
*   *
*****
```

````python
n = int(input("Enter the no of rows you want:- "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print('*', end='')
        else:
            print(' ', end='')
    print()
````

### Solution for 'CSV to JSON conversion'

````python
import csv
import json

# 1. Create a dummy students.csv file
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Age", "Grade"])
    writer.writerows([
        ["Alice", 20, "A"],
        ["Bob", 22, "B"],
        ["Charlie", 21, "A"]
    ])
print("Created dummy students.csv")
````

````python
# 2. Read students.csv using DictReader and convert to list of dictionaries
# 3. Write that list into a file called students_converted.json using json.dump

students_data = []
with open("students.csv", "r") as csv_file:
    reader = csv.DictReader(csv_file)
    for row in reader:
        students_data.append(row)

with open("students_converted.json", "w") as json_file:
    json.dump(students_data, json_file, indent=4)

print("Conversion complete: students.csv -> students_converted.json")
````

````python
# 4. Read the JSON file back and print just the names

with open("students_converted.json", "r") as json_file:
    data = json.load(json_file)
    print("\nNames from JSON file:")
    for student in data:
        print(student["Name"])
````

## Difference between element multiplication (*) and matrix multiplication(@)

### Element-wise vs. Matrix Multiplication

**Element-wise multiplication (`*`)** performs multiplication on corresponding elements of two arrays. Both arrays must have compatible shapes (either identical or broadcastable).

````python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

ele_mul = arr1 * arr2
print("Array 1:\n", arr1)
print("Array 2:\n", arr2)
print("\nElement-wise product:\n", ele_mul)
````

**Matrix multiplication (`@`)** follows the rules of linear algebra. For two matrices A (m x n) and B (n x p), their product C (m x p) is defined if the number of columns in A equals the number of rows in B.

````python
mat1 = np.array([[1, 2], [3, 4]])
mat2 = np.array([[5, 6], [7, 8]])

matrix_product = mat1 @ mat2
print("Matrix 1:\n", mat1)
print("Matrix 2:\n", mat2)
print("\nMatrix product :\n", matrix_product)

mat_A = np.array([[1, 2, 3], [4, 5, 6]])  # 2x3
mat_B = np.array([[7, 8], [9, 10], [11, 12]])  # 3x2

mat_product1 = mat_A @ mat_B
print("\nMatrix A:\n", mat_A)
print("Matrix B:\n", mat_B)
print("\nMatrix product :\n", mat_product1)
````

## Basic Statistics

Now move from understanding the dataset to understanding the numbers.

* sum()
* mean()
* median()
* min()
* max()
* describe()

```python
sales=df["Sales"]      # keep the column in a variable so we do not type it again and again
print(f"Average Sales :- {sales.mean()}")   # average sales
print(f"Total Sales :- {sales.sum()}")      # total sales
print(sales.min())      # min sales value
print(sales.median())   # middle value when data is sorted
print(sales.max())      # max sales value
```

Mean is much bigger than median because a few very big orders pull the mean up. When data has extreme values median is the better typical value.

```python
df["Profit"].describe()    # statistical summary
```

```python
df[["Sales","Quantity","Discount","Profit"]].describe()   # all important numeric columns together
```

`max()` gives only the value. `idxmax()` gives the **index of the row** which has the maximum value. Then we can use `.loc` to see that row.

```python
big=df["Sales"].idxmax()     # index of the row with highest sales
print(big)
df.loc[big,["Order ID","Category","Sub-Category","Sales","Profit"]]
```

```python
df.loc[df["Sales"].idxmin()]     # idxmin gives the row with minimum value
```

# 3. Filtering Rows (Boolean Masking)

Filtering means keeping only the rows where a condition is True.

1. A condition like `df["Profit"]<0` gives True/False for every row. This is called a mask.
2. `df[mask]` keeps only the True rows.

| Operator | Meaning |
|---|---|
| `==`  `!=`  `>`  `>=`  `<`  `<=` | compare |
| `&` | AND (both conditions true) |
| `\|` | OR (any one true) |
| `~` | NOT |

Remember: `==` is for comparing, `=` is for assigning. Use `&` `|` and not `and` `or`. Each condition must be inside brackets.

```python
df["Profit"]<0      # this is the mask, True where the order made loss
```

```python
big_orders=df[df["Sales"]>1000]     # only True rows are kept
print(big_orders.shape)
big_orders.head()
```

```python
df[df["Region"]=="Central"].head()
```

### More than one condition

```python
df[(df["Profit"]<0) & (df["Category"]=="Technology")].head()
```

```python
df[(df["Region"]=="West") & (df["Quantity"]>=10)].head()
```

```python
# OR : same day delivery or very high discount
df[(df["Ship Mode"]=="Same Day") | (df["Discount"]>=0.5)].shape
```

```python
df[~(df["Region"]=="East")]["Region"].unique()    # NOT east
```

### isin, between and str.contains

```python
df[df["Region"].isin(["East","West"])].shape     # value is any one from the list
```

```python
df[df["Sales"].between(1000,5000)].shape         # 1000 <= Sales <= 5000, both ends included
```

```python
df[df["Product Name"].str.contains("Chair")].head(3)    # for text use .str first
```

# 4. Indexing: .loc and .iloc

Every dataframe has an index (row labels). By default it is 0,1,2,...

| | .loc | .iloc |
|---|---|---|
| uses | labels (index name, column name) | position (0,1,2...) |
| end of slice | included | not included (like python list) |
| boolean condition | can be used | not used |

A boolean condition can be used inside `.loc` to filter rows and choose columns in one step: `df.loc[condition,[columns]]`

```python
df.iloc[0:5]       # rows at position 0 to 4, 5 is not included
```

```python
df.iloc[0:3,0:4]   # first 3 rows and first 4 columns
```

```python
df.loc[0:4,["Order ID","Sales"]]    # rows with label 0 to 4, here 4 is included
```

```python
df.loc[(df['Sales']>=7000)&(df['Category']=='Technology'),['Row ID','Customer ID','Customer Name','Category','Sales']]
```

```python
df.loc[(df['Profit']<0)& (df['Category']=='Furniture'),['Order ID','Category','Sales','Discount','Profit']].head()
```

### set_index() and reset_index()

`set_index` makes a column the row labels. We keep the result in a new variable `df1` so `df` stays same.

```python
df1=df.set_index('Row ID')    # used for setting a particular column as index column
df1.head(3)
```

```python
print(df1.loc[11].name)      # label 11 means row with Row ID 11
print(df1.iloc[11].name)     # position 11 is the 12th row, so Row ID is 12
```

```python
df1.reset_index().head(3)    # index comes back as a normal column
```

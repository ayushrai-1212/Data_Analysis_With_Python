# Write a code to find leap year?

````python
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
````

#Electricity bill calculator

100 units 5 rupees
200 units 7 rupees
350 10 rupees
450 15 rupees

````python
units = int(input("Enter the number of units consumed: "))
bill = 0

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
elif units <= 350:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)
elif units <= 450:
    bill = (100 * 5) + (100 * 7) + (150 * 10) + ((units - 350) * 15)
else:
    bill = (100 * 5) + (100 * 7) + (150 * 10) + (100 * 15) + ((units - 450) * 20) # Assuming 20 for units above 450
print(f"Your electricity bill is: ${bill}")
````

## Pattern Printing Exercises

Let's explore various pattern printing problems using loops.

### Q1. Right-Angled Triangle (Asterisks)

Print a right-angled triangle pattern of asterisks:

```
*
**
***
****
```

````python
rows = int(input("Enter the no of rows you want:- "))
for i in range(1, rows + 1):
    for j in range(i):
        print('*', end='')
    print()
````

### Q2. Pyramid (Asterisks)

Print a pyramid pattern of asterisks:

```
   *
  ***
 *****
*******
```

````python
rows = int(input("Enter the no of rows you want:- "))
for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
````

### Q3. Diamond (Asterisks)

Print a diamond pattern of asterisks:

```
   *
  ***
 *****
*******
 *****
  ***
   *
```

````python
rows = int(input("Enter the no of rows you want:- "))
# Upper part of the diamond
for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
# Lower part of the diamond
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
````

### Q4. Hollow Square (Asterisks)

Print a hollow square pattern of asterisks:

```
*****
*   *
*   *
*   *
*****
```

````python
n = int(input("Enter the no of rows you want:- "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print('*', end='')
        else:
            print(' ', end='')
    print()
````



## break = leave the loop completely

````python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
````

### Question 4

Does 6 get printed?

No. break immediately terminates the loop.

## continue = skip the current iteration

````python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````



- **break → leave the loop**
- **continue → skip this round**
